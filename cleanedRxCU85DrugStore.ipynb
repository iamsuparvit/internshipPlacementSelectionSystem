```python
input_file = './RxCU85DrugStore.csv'
df = pd.read_csv(input_file)
df
```

```python
df = df.dropna(axis=0, how='all')
df = df.dropna(axis=1, how='all')
df
```

```python
df = df.rename(columns={
   'Unnamed: 0':'company',
   'Unnamed: 1':'branch',
   'ที่อยู่':'address',
   'ผลัดที่ 1 (4 มิ.ย. - 1 ก.ค. 69)':'sex_require1',
   'Unnamed: 3':'seat1',
   'ผลัดที่ 2 (2 - 31 ก.ค. 69)':'sex_require2',
   'Unnamed: 5':'seat2',
   'ผลัดที่ 3 (เสาร์ อาทิตย์ 1 ส.ค. - 4 ต.ค. 69)':'sex_require3',
   'Unnamed: 7':'seat3'
})
```

```python
df = df.drop(index=0, errors='ignore')
df
```

```python
company_prefix = {
    'ศบส. กองเภสัชกรรม': 'CEN',
    'ร้านยากรุงเทพ': 'BKK',
    'ร้านยา Boots': 'BTS',
    'โรงพยาบาลตำรวจ': 'HPT',
    'จามฯ 9': 'CHA',
    'ร้านยา Pure': 'PUR',
    'ร้านยา Fascino': 'FCN',
}

# รันเลขจาก 1 ไป n
df['company'] = df['company'].ffill()

counters = {prefix: 1 for prefix in company_prefix.values()}
df['code'] = None

for idx in df.index:
    company = df.loc[idx, 'company']
    if company in company_prefix:
        prefix = company_prefix[company]
        df.loc[idx, 'code'] = f'{prefix}{counters[prefix]:02d}'
        counters[prefix] += 1

df
```

```python
df = df[['code','branch','sex_require1','seat1','sex_require2','seat2','sex_require3','seat3']]
df
```

```python
for i in range(1,4):
    df.loc[:, f'seat{i}'] = df[f'seat{i}'].fillna(0)
```

```python
sex_map = {
    'ไม่ระบุเพศ': 'Both',
    'ช เท่านั้น': 'Male',
    'ช': 'Male',
    'ญ เท่านั้น': 'Female',
    'ญ': 'Female',
    'ไม่รับ': None
}

for j in range(1, 4):
    col = f'sex_require{j}'
    df[col] = df[col].apply(lambda x: sex_map.get(str(x).strip(), None) if pd.notna(x) and str(x).strip() != '' else None)

df
```
